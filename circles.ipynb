```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("results.csv")
```

```python
import matplotlib.pyplot as plt

# Разделяем на два типа прямоугольника
df_narrow = df[df["rect"] == "Accurate"]
df_wide   = df[df["rect"] == "Wider"]

A_ref = df["exactArea"].iloc[0]   # Берём точную площадь

# ---------- График 1: приближённая площадь ----------
plt.plot(df_narrow["N"], df_narrow["estArea"], label="точный прямоугольник")
plt.plot(df_wide["N"], df_wide["estArea"], label="широкий")
plt.axhline(A_ref, linestyle="--", label="точная площадь")

plt.xlabel("N")
plt.ylabel("Площадь")
plt.title("Приближённая площадь vs N")
plt.legend()
plt.grid(True)
plt.show()

# ---------- График 2: относительная ошибка ----------
plt.plot(df_narrow["N"], df_narrow["relError"], label="точный прямоугольник")
plt.plot(df_wide["N"], df_wide["relError"], label="широкий")

plt.xlabel("N")
plt.ylabel("Относительная ошибка")
plt.title("Ошибка vs N")
plt.legend()
plt.grid(True)
plt.show()
```
